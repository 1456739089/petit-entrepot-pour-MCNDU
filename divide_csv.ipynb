```python
#read csv
import pandas as pd
df_origin = pd.read_csv(
    "MCNDU_final_dataset.csv", 
    sep=';'
)
df_origin.head()
```

```python
list(df_origin.columns)
```

```python
# 6 个筛分标志列（0/1）
flag_cols = ['near_A1',
 'near_A3',
 'near_A4',
 'near_A6a',
 'near_A6b',
 'near_A13']
feature_cols = ['Occupation',
 'Debit',
 'Arc Name',
 'Arc ID',
 'Datetime',
 'Trafic',
 'Latitude',
 'Longitude']
rename_dict = {
    "Occupation": "Taux d'occupation",
    "Debit": "Débit horaire",
    "Datetime": "Date et heure de comptage",
}
all_dfs = []   # 用来存所有子 df
for col in flag_cols:
    df_sub = df_origin.loc[df_origin[col] == 1, feature_cols]
    df_sub = df_sub.rename(columns=rename_dict)
    df_sub.to_csv(f"{col}_is_1.csv", index=False)
    all_dfs.append(df_sub)

# 循环结束后合并
df_all = pd.concat(all_dfs, ignore_index=True)
df_all.to_csv("all_flags_merged.csv", index=False)
```
