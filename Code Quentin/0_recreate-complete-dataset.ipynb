#### Exploration de correspondances entre les stations

On a vu dans la partie 4. que 20% des stations du dataset comptent le taux d'occupation mais pas le débit horaire, et 17% comptent le débit horaire mais pas le taux d'occupation.

Pour nettoyer le dataset et avoir des données complètes, on peut essayer de voir si certaines stations de comptage du débit sont positionnées à côté de certaines stations de comptage du taux d'occupation. Alors, si les deux stations prennent des mesures dans la même heure, on peut fusionner les données pour récupérer une ligne complète.

```python
# Distance entre deux points GPS

def dist_haversine(lat1, lon1, lat2, lon2):
    R = 6371000
    phi1, phi2 = np.radians(lat1), np.radians(lat2)
    dphi = phi2 - phi1
    dlambda = np.radians(lon2 - lon1)
    
    a = np.sin(dphi/2)**2 + np.cos(phi1)*np.cos(phi2)*np.sin(dlambda/2)**2
    return 2*R*np.arcsin(np.sqrt(a))
```

```python
def find_matching_stations(
    df_debit_only,
    df_occupation_only,
    distance_threshold=50,
    same_nodes=True,
    close_enough=True,
    similar_name=True
):
    """
    Trouve des paires (débit, taux) pour des mesures effectuées au même instant.
    """

    matching_pairs = []

    # Pour éviter de faire matcher plusieurs fois la même ligne de débit
    used_debit_indices = set()

    # Boucle sur les lignes avec taux d'occupation uniquement
    for occ_idx, row_occ in df_occupation_only.iterrows():
        time_i = row_occ["Date et heure de comptage"]

        # Candidats = seules les lignes de débit à la même date/heure
        candidates = df_debit_only[
            df_debit_only["Date et heure de comptage"] == time_i
        ]

        for deb_idx, row_deb in candidates.iterrows():
            # Si cette ligne de débit a déjà été utilisée dans une autre paire, on la saute
            if deb_idx in used_debit_indices:
                continue

            condition_results = []

            # Condition 1 : partage de nœuds
            if same_nodes:
                cond_same_nodes = (
                    row_occ["Identifiant noeud amont"] == row_deb["Identifiant noeud amont"] or
                    row_occ["Identifiant noeud aval"] == row_deb["Identifiant noeud aval"] or
                    row_occ["Identifiant noeud aval"] == row_deb["Identifiant noeud amont"] or
                    row_occ["Identifiant noeud amont"] == row_deb["Identifiant noeud aval"]
                )
                condition_results.append(cond_same_nodes)

            # Condition 2 : distance géographique
            if close_enough:
                dist = dist_haversine(
                    row_occ["latitude"], row_occ["longitude"],
                    row_deb["latitude"], row_deb["longitude"]
                )
                cond_close_enough = dist <= distance_threshold
                condition_results.append(cond_close_enough)

            # Condition 3 : libellés similaires
            if similar_name:
                lib_occ = str(row_occ["Libelle"])
                lib_deb = str(row_deb["Libelle"])
                cond_similar_name = lib_occ.split("_")[0] == lib_deb.split("_")[0]
                condition_results.append(cond_similar_name)

            # Si aucune condition n'est activée, on ne matche rien
            if not condition_results:
                continue

            # On matche si au moins une des conditions activées est vraie
            if any(condition_results):
                matching_pairs.append((deb_idx, occ_idx))
                used_debit_indices.add(deb_idx)
                # On passe au point de taux suivant (on force un matching 1–1)
                break

    return matching_pairs
```

```python
def merge_debit_occupation(
    df_debit_only,
    df_occupation_only,
    distance_threshold=50,
    same_nodes=True,
    close_enough=True,
    similar_name=True
):
    """
    Retourne un DataFrame fusionné à partir de df_debit_only et df_occupation_only
    en utilisant les paires trouvées par find_matching_stations.
    """

    # Copies pour ne pas modifier les originaux
    debit = df_debit_only.copy()
    occ = df_occupation_only.copy()

    # 1) Trouver les paires
    matching_pairs = find_matching_stations(
        debit,
        occ,
        distance_threshold=distance_threshold,
        same_nodes=same_nodes,
        close_enough=close_enough,
        similar_name=similar_name,
    )

    # 2) Appliquer les appariements : remplir le taux d'occupation côté débit
    matched_occ_indices = set()

    for deb_idx, occ_idx in matching_pairs:

        taux_from_occ = occ.at[occ_idx, "Taux d'occupation"]
        taux_in_deb = debit.at[deb_idx, "Taux d'occupation"]

        # On ne remplit que si NaN côté débit
        if pd.isna(taux_in_deb) and not pd.isna(taux_from_occ):
            debit.at[deb_idx, "Taux d'occupation"] = taux_from_occ

        matched_occ_indices.add(occ_idx)

    occ_remaining = occ.drop(index=list(matched_occ_indices))

    df_merged = pd.concat([debit, occ_remaining], axis=0)

    if "Date et heure de comptage" in df_merged.columns:
        df_merged = df_merged.sort_values(by="Date et heure de comptage", ascending=False)

    df_merged = df_merged.reset_index(drop=True)

    return df_merged
```

```python
def build_full_dataset(
    df,
    df_debit_only,
    df_occupation_only,
    distance_threshold=50,
    same_nodes=True,
    close_enough=True,
    similar_name=True
):
    """
    Construit un dataset complet à partir :
      - du df original
      - des sous-ensembles df_debit_only et df_occupation_only
    """

    # 1) Lignes complètes dans df
    df_complete = df[
        df["Débit horaire"].notna() & df["Taux d'occupation"].notna()
    ].copy()

    # 2) Fusion des df_debit_only et df_occupation_only
    df_partial_merged = merge_debit_occupation(
        df_debit_only,
        df_occupation_only,
        distance_threshold=distance_threshold,
        same_nodes=same_nodes,
        close_enough=close_enough,
        similar_name=similar_name,
    )

    df_merged = pd.concat([df_complete, df_partial_merged], axis=0)

    if "Date et heure de comptage" in df_merged.columns:
        df_merged = df_merged.sort_values(
            by="Date et heure de comptage",
            ascending=False
        )

    return df_merged
```
