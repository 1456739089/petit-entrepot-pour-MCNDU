```python
import geopandas as gpd
from shapely.geometry import Point

ztl = gpd.read_file("/Users/ENPC/cours 3/3 MCNDU/projet/ztl.geojson") 

import pandas as pd
df = pd.read_csv("/Users/ENPC/cours 3/3 MCNDU/projet/points.csv")  

df["geometry"] = df.apply(lambda row: Point(row["longitude"], row["latitude"]), axis=1)
points = gpd.GeoDataFrame(df, geometry="geometry", crs="EPSG:4326")

# projection
points = points.to_crs(ztl.crs)

points["in_ztl"] = points.within(ztl.unary_union)

points.head()
```
